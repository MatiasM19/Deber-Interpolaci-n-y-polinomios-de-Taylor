```python
import numpy as np

def taylor_sin(x, n):
    x = np.asarray(x, dtype=float)
    t = x.copy()
    s = t.copy()

    for k in range(n // 2):
        t = t * (-x**2) / ((2*k+2)*(2*k+3))
        s = s + t

    return s

print(f"{taylor_sin(np.pi, 9):.6f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2, 2, 2001)

plt.plot(x, np.sin(x), label="sin(x)")

for n in [1, 3, 5, 7, 9]:
    plt.plot(x, taylor_sin(x, n), label=f"P{n}(x)")

plt.ylim(-2, 2)
plt.grid()
plt.legend()
plt.title("Serie de Taylor de sin(x)")
plt.show()
```

```python
import numpy as np

grados = range(1, 22, 2)

x1 = np.linspace(-np.pi, np.pi, 2001)
x2 = np.linspace(-np.pi/4, np.pi/4, 2001)

errores1 = []
errores2 = []

for n in grados:
    errores1.append(np.max(np.abs(np.sin(x1) - taylor_sin(x1, n))))
    errores2.append(np.max(np.abs(np.sin(x2) - taylor_sin(x2, n))))

print("Intervalo [-pi, pi]:")
print(errores1)

print("Intervalo [-pi/4, pi/4]:")
print(errores2)
```

```python
import numpy as np

grados = range(1, 22, 2)

x1 = np.linspace(-np.pi, np.pi, 2001)
x2 = np.linspace(-np.pi/4, np.pi/4, 2001)

errores1 = []
errores2 = []

for n in grados:
    errores1.append(np.max(np.abs(np.sin(x1) - taylor_sin(x1, n))))
    errores2.append(np.max(np.abs(np.sin(x2) - taylor_sin(x2, n))))

plt.plot(list(grados), errores1, 'o-', label="[-pi, pi]")
plt.plot(list(grados), errores2, 'o-', label="[-pi/4, pi/4]")

plt.yscale("log")
plt.xlabel("Grado n")
plt.ylabel("Error máximo")
plt.title("Errores máximos de Taylor")
plt.grid()
plt.legend()
plt.show()
```

#### Parte 3

```python
import numpy as np

def serie_geometrica(x, N):
    t = 1.0
    s = t

    for k in range(N):
        t = t * (-x**2)
        s = s + t

    return s


r1 = serie_geometrica(0.9, 40)
r2 = serie_geometrica(1.1, 40)

print(f"S40(0.9) = {r1:.6f}")
print(f"S40(1.1) = {r2:.6f}")
```

```python
import matplotlib.pyplot as plt

x = np.linspace(-1.5, 1.5, 2001)

plt.plot(x, 1/(1+x**2), label="Funcion real")

for N in [2, 5, 10, 20]:
    plt.plot(x, serie_geometrica(x, N), label=f"S{N}(x)")

plt.ylim(-0.5, 1.5)
plt.grid()
plt.legend()
plt.title("Serie de Maclaurin de 1/(1+x²)")
plt.show()
```

```python
x = -20
t = 1.0
mayor = abs(t)

for k in range(1, 101):
    t = t * x / k
    if abs(t) > mayor:
        mayor = abs(t)

print("Mayor término =", mayor)
```
